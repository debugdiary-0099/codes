```python

# Experiment 1: Feature Engineering on Titanic Dataset

# Step 1: Install required libraries
%pip install numpy pandas matplotlib seaborn scikit-learn

# Step 2: Import required libraries
import numpy as np
import pandas as pd
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler

# Step 3: Load the Titanic dataset
df = sns.load_dataset("titanic")

# Step 4: Display first five records
print("First Five Records:")
print(df.head())

# Step 5: Display dataset shape
print("\nDataset Shape:")
print(df.shape)

# Step 6: Display column names
print("\nColumn Names:")
print(df.columns)

# Step 7: Display dataset information
print("\nDataset Information:")
df.info()

# Step 8: Display summary statistics
print("\nSummary Statistics:")
print(df.describe())

# Step 9: Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 10: Select relevant columns
data = df[["survived", "pclass", "sex", "age",
           "sibsp", "parch", "fare", "embarked"]].copy()

# Step 11: Handle missing values
data["age"] = data["age"].fillna(data["age"].median())
data["fare"] = data["fare"].fillna(data["fare"].median())
data["embarked"] = data["embarked"].fillna(
    data["embarked"].mode()[0]
)

# Step 12: Check missing values after handling
print("\nMissing Values After Handling:")
print(data.isnull().sum())

# Step 13: Encode categorical variables
data = pd.get_dummies(
    data,
    columns=["sex", "embarked", "pclass"],
    dtype=int
)

# Step 14: Scale numerical features
scaler = MinMaxScaler()
num_cols = ["age", "sibsp", "parch", "fare"]

data[num_cols] = scaler.fit_transform(data[num_cols])

# Step 15: Display cleaned dataset
print("\nCleaned Dataset:")
print(data.head())

# Step 16: Verify final dataset
print("\nFinal Dataset Shape:")
print(data.shape)

print("\nFinal Missing Values:")
print(data.isnull().sum())

print("\nFinal Dataset Information:")
data.info()
# Check duplicate records
print("\nDuplicate Records:", df.duplicated().sum())

# Check unique values in each column
print("\nUnique Values:")
print(df.nunique())

print("\nFeature Engineering Completed Successfully!")
```
